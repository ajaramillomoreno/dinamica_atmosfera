```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
## Velocidad angular de rotación de la Tierra en rad s^-1
Omega = 7.292e-5

# Constantes 
Pi = np.pi


def f_coriolis(latitud):
    latitud_rad = latitud*Pi/180.0
    return 2.0*Omega*np.sin(latitud_rad)
```

```python
X0 = 0
Y0 = 0
U0 = 1000 # m/s
V0 = 0
t0 = 0
latitud = 43

f = f_coriolis(latitud)

delta_t = 10*60 #10 minutos en segundos
time_max = 3*60*60 # 3 horas de vuelo


t = [t0]
X = [X0]
Y = [Y0]
U = [U0]
V = [V0]

for ti in range(delta_t,time_max+delta_t,delta_t):
    t.append(t[-1] + delta_t)
    X.append(X[-1] + U[-1]*delta_t)
    Y.append(Y[-1] + V[-1]*delta_t)
    U.append(U[-1] + f*V[-1]*delta_t)
    V.append(V[-1] -f*U[-1]*delta_t)
    
    
    
df = pd.DataFrame({'time':t,'X':X,'Y':Y,'U':U,'V':V})
df = df.set_index('time')
df
```

```python
fig, ax = plt.subplots(figsize=(15,5))
ax.plot(df['X'].values/1000, df['Y'].values/1000, ls='-', color='red', lw = 2.0) 


ax.axhline(0,ls='--', color='grey',lw=1.0)
ax.set_xlabel('X [km]')
ax.set_ylabel('Y [km]')
ax.set_xlim(0)
ax.set_ylim(-8000,8000)
plt.show()
```
